```python
import os
from datasets import load_dataset
from tqdm import tqdm

# Define o caminho de destino
OUTPUT_DIR = "/mnt/study-data/dcarvalho/datasets/celebahq"
os.makedirs(OUTPUT_DIR, exist_ok=True)

print("Baixando e carregando o dataset do Hugging Face...")
# Carrega as divisões 'train' e 'validation'
dataset = load_dataset("mattymchen/celeba-hq")
print("Dataset carregado com sucesso!")
```

```python
count = 0

# Percorre os dois splits (train e validation)
for split in ['train', 'validation']:
    print(f"\nSalvando imagens do split '{split}'...")
    for item in tqdm(dataset[split]):
        image = item['image']
        
        # Define o caminho e nome da imagem: img_00000.jpg, img_00001.jpg, ...
        image_path = os.path.join(OUTPUT_DIR, f"img_{count:05d}.jpg")
        
        # Salva a imagem em alta resolução
        image.save(image_path, format="JPEG", quality=95)
        count += 1

print(f"\nConcluído! Foram salvas {count} imagens em: {OUTPUT_DIR}")
```
